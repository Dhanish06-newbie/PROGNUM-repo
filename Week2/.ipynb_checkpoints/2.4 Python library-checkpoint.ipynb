```python
import math


x = float(input("Enter the value for x: "))
y = float(input("Enter the value for y: "))


angle_rad = math.atan2(y, x)
angle_deg = math.degrees(angle_rad)

print(f"\nThe angle between the x-axis and the point ({x}, {y}) is {angle_deg:.2f} degrees.")


if x > 0 and y > 0:
    quadrant = "Quadrant I"
elif x < 0 and y > 0:
    quadrant = "Quadrant II"
elif x < 0 and y < 0:
    quadrant = "Quadrant III"
elif x > 0 and y < 0:
    quadrant = "Quadrant IV"
elif x == 0 and y == 0:
    quadrant = "the Origin"
else:
    quadrant = "one of the Axes"

print(f"The point belongs to: {quadrant}")
```

```python
# Define the lists
night1 = [1, 3, 2, 4, 0, 3, 7, 5, 2, 9, 2, 4, 3]
night2 = [0, 0, 2, 4, 2, 4, 3, 2, 1, 1, 1, 2, 0]


set1 = set(night1)
set2 = set(night2)


common_values_set = set1.intersection(set2)


intersection_list = list(common_values_set)
print(f"Intersection list: {intersection_list}")


is_5_in_intersection = 5 in intersection_list
print(f"Is 5 in the intersection list? {is_5_in_intersection}")


is_5_in_set1 = 5 in set1
print(f"Is 5 in the set version of night1? {is_5_in_set1}")
```

```python
# Question 19


import numpy as np
data = np.random.normal(loc=0.0, scale=1.4, size=100000)

calc_std = np.std(data)

print(f"Target Std Dev: 1.4")
print(f"Calculated Std Dev: {calc_std}")
```

```python
# Question 21


a = 1
b = 2
count = 2  # We already have the first two terms


while count < 100:
    # Update values: b becomes the sum, a becomes the old b
    a, b = b, a + b
    count += 1

print(f"The 100-th term is: {b}")
```
