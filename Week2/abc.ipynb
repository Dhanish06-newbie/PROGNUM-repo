```python
import math

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

if a == 0:
    if b != 0:
        x = -c / b
        print(f"One solution (linear): {x}")
    else:
        print("No solution.")
else:
    D = b**2 - 4*a*c
    if D > 0:
        x1 = (-b + math.sqrt(D)) / (2*a)
        x2 = (-b - math.sqrt(D)) / (2*a)
        print(f"Two solutions: {x1} and {x2}")
    elif D == 0:
        x = -b / (2*a)
        print(f"One solution: {x}")
    else:
        print("No real solutions could be found.")
```
