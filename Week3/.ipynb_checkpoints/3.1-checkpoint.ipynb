```python
#Question 3.2
import math

masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']

G = 6.67430e-11
escape_velocities = {}

for name, m, r in zip(names, masses, radii):
    r_meters = r * 1000
    ve = math.sqrt((2 * G * m) / r_meters)
    escape_velocities[name] = ve
    print(f"{name:8} ; Mass: {m:.2e} kg ; Escape Velocity: {ve/1000:.2f} km/s")

print("\nFinal Dictionary:")
print(escape_velocities)
```

```python
#Question 3.7
import math

def angular_size(s, d):
    """
    s: linear size of the object
    d: distance to the object
    """
    theta_rad = 2 * math.atan(s / (2 * d))
    theta_deg = math.degrees(theta_rad)
    
    return theta_deg

# Eg: The Moon
# s = 3474 km, d = 384400 km
print(f"The agular size of the moon is : {angular_size(3474, 384400)}")
```

```python
#Question 3.10
import math

def cylinder(r, h):
    area = 2 * math.pi * r * (r + h)
    volume = math.pi * (r ** 2) * h
    return area, volume

radius = 5
height = 10
a, v = cylinder(radius, height)
ratio = a / v

print(f"Area: {a}")
print(f"Volume: {v}")
print(f"Ratio: {ratio}")
```

```python
#Question 3.17

import numpy as np

def stdev(data):

    x = np.array(data)
    n = len(x)
    
    mu = np.sum(x) / n
    sum_sq_diff = np.sum((x - mu)**2)    
    sigma = np.sqrt(sum_sq_diff / n)
    
    return sigma


test_data = [10, 2, 38, 23, 38, 23, 21]
my_result = stdev(test_data)
np_result = np.std(test_data)

print(f"My Function: {my_result}")
print(f"NumPy's Function: {np_result}")
```

```python
#Question 3.18
import numpy as np

# Distance (Mpc) and Velocity (km/s)
x = np.array([18.49, 35.48, 35.48, 11.34, 7.12, 9.82, 23.66, 44.0, 66.7, 52.4, 55.0, 25.6, 26.19])
y = np.array([2094.75, 2320.87, 1224.9, 872.1, 432.2, 751.0, 1410.0, 2700.0, 5292.0, 2550.0, 5253.0, 2419.0, 2158.0])

x_bar = np.mean(x)
y_bar = np.mean(y)
#y = mx+c
m = np.sum((x - x_bar) * (y - y_bar)) / np.sum((x - x_bar)**2)

c = y_bar - m * x_bar

print(f"Slope (m): {m:.4f}")
print(f"Intercept (c): {c:.4f}")
```

```python
#Question 3.22
import matplotlib.pyplot as plt
import numpy as np
import math

#Using range
x_range = list(range(-4, 6))
y_range = [math.cosh(x) for x in x_range]

# Using NumPy arange() for the same values
x_numpy = np.arange(-4, 6)
y_numpy = np.cosh(x_numpy)


plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, 'bo-', label='Catenary (range)')
plt.plot(x_numpy, y_numpy, 'rx--', label='Catenary (arange)', alpha=0.5)


plt.title('Catenary Curve: $y = \\cosh(x)$', fontsize=16)
plt.xlabel('x-axis', fontsize=14)
plt.ylabel('y-axis', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)


plt.savefig('catenary_plot.png')
```
