```python
#5.1
clusters = [
    "Virgo Cluster",
    "Fornax Cluster",
    "Hercules Cluster",
    "Coma Cluster",
    "Norma Cluster"
]

file_name = "galaxy_clusters.txt"

with open(file_name, "w") as f:
    for cluster in clusters:
        f.write(cluster + "\n")
        
with open(file_name, "r") as file:
    for i, line in enumerate(file, 1):
        names = line.strip()
        print(f"Galaxy cluster {i} is {names}")
f.close()
```
