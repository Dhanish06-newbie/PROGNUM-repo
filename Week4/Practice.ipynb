```python
import numpy as np
z= 6+7j
length = np.abs(z)
print ("Length of vector :",length)
degs = np.angle(z,deg=True)
print (degs,"degrees")
print ("z is complex?", np.iscomplex(z))
print(f"3 times z is : =,{3*z}")

def polar2complex(lengths, angles):
    """Convert length and angle to a complex number.
       The function arguments can be NumPy arrays"""
    return lengths * np.exp(1j*angles)

angle = np.radians(degs)
z = polar2complex(length, angle)
print(f"z={z:.1f}")
    
```

```python
def scaling(value, bscale=1, bzero=0):
    """ Scale a value with bscale and add offset bzero"""
    result1 = bscale*value + bzero
    result2 = bscale * value
    return result1, result2

val = 3
r1, r2 = scaling(val, 2, 0.01)
print(r1, r2)
```
