```python
import numpy as np

x = np.nan
print(x == np.nan)
print(x is np.nan)
print(np.isnan(x))
print(~np.isnan(x))
```

```python
import numpy as np

x = np.array([1.0, 2, np.nan, -2e3, np.nan, 7e333])  # Just an array with some NaN's
print("x=                       ", x)
print("isnan(x)=                ", np.isnan(x))
print("The indices where True :", np.where(np.isnan(x))[0])   
```

```python
import numpy as np
x= np.array([3,4,2,np.nan,3,2.0,-1])
idx_max = np.argmax(x) #Gives the index of 1st nan encountered'''
idx_nanargmax = np.nanargmax(x) #Ignores all Nan to find the true max'''
print (idx_argmax)
print (idx_nanargmax)
```

```python
import numpy as np

counts = np.array([123432, np.nan, 22333, 344433, np.inf, 222222])

#part 1
filter_mask = np.isfinite(counts)
filtered_counts = counts[filter_mask]
print("Filtered Array:", filtered_counts)

#part 2
masked_counts = np.ma.masked_invalid(counts)

print("Masked Array:")
print(masked_counts)
```
